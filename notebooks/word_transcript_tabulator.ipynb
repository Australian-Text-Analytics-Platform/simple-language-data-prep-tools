# About The Options

## Uploading a matched spreadsheet

If you've already run this process and created a spreadsheet, but now need to rerun the process, you can upload your existing spreadsheet and have it merged in.

Any new sheets you've added will be retained as is. Any new columns you've added as part of extending information in the sheets created by this tool will be merged in. Rows will be merged based on the unique identifier columns for each sheet: rows that can't be matched to anything existing will be appended at the end to prevent loss of information.

This is designed to help you if you want the `.docx` format of your transcript to be the version of record. If you identify issues such as a variation in how the same speaker was identified, you can correct this in your word transcript, reupload your edited transcripts and the edited spreadsheet, and produce a new and corrected spreadsheet. 

## Speaker separator

Specifies how the speaker code and the text are separated in your transcripts: the default option (colon-tab) is a common convention while (colon-space) is sometimes used as well. The (colon-anywhitespace) option can be useful when your transcripts are not consistent, but can lead to incorrect identification of speaker codes.

## Assign previous speaker when no speaker identified

Sometimes consecutive turns/utterances/units-of-talk by the same speaker are not given explicit speaker codes: by convention the last speaker is assumed to have continued talking. Ticking this box will apply this convention. When a speaker code appears to be missing, the last valid speaker code will be applied to this unit. 

Turning this on can lead to incorrect speaker code assignment when speaker codes are inconsistently separated from the transcript. It is recommended to try without this option first, unless you already know that this convention applies in the transcripts you're working with.

# About The Output

The output of this tool is a single spreadsheet in Microsoft Excel (`.xlsx`) format. This spreadsheet treats each paragraph (approximately whenever you insert a line break with the `return/enter` key) in the Word file as a single unit of text, which is assumed to have a single associated speaker. 

This spreadsheet contains five individual sheets: `turn`, `speaker_code`, `segment`, `transcript_file`, and `provenance`.

## Turn

The `turn` sheet indicates the individual units of text extracted from the file: this is the core transcribed text.

Each row in this sheet corresponds to one unit of text (this might be a turn/utterance/question/answer/response/comment depending on your disciplinary context and approach to transcription). 

Each row is identified by the source file it came from, and a `turn_no` that starts at 1 with the first unit of text extracted from the file. 

The transcription column contains the extracted text

## Speaker Code

The `speaker_code` sheet indicates the summary of extracted speakers from the transcripts.

Each row corresponds to one unique speaker code per uploaded file. The number of turns assigned to each speaker_code are indicated. Turns out without identified speakers are marked with a blank cell. 

You can use this table to check the consistency of speaker codes across the set of transcripts. This will highlight inconsistencies, for example where the same speaker might be `AB` in one line of a transcript and `Ab` in another.

## Segment

The `segment` sheet will be useful if your transcripts have used blank lines to indicate higher level structures/sections in your transcripts. If your transcripts don't have any segments this sheet will have a single line per transcript file and can be safely ignored.

## Transcript File

The `transcript_file` sheet gives you an overview of the size of each of the uploaded transcripts. This provides provenance information to help you confirm you have included and processed the correct number of transcripts. It also helps understand your transcripts as a whole: unusually long or short files may indicate issues with the transcript (or recording) that need further investigation.


## Provenance 

The `provenance` table provides some basic information about this tool, including the DOI, the version number and a suggested citation. It also includes a basic summary of the provided files used to generate this spreadsheet. 



# Word File Processing Notes

Tables are completely skipped during this process - only running paragraphs are included. If you want to exclude text from the spreadsheet (such as an informational header) you can wrap it in a table.

Completely blank lines (pressing `enter/return` twice in a row) are treated as "segment" markers. If your transcripts indicate some kind of extra structure besides the flow of transcribed text you can use this to mark segments for annotation. As one example of how this can be used, if you have a semi-structured interview with a core set of questions/topics across participants, you can use this to delimit where transitions between these main questions/topics occur.

Speaker codes are identified by looking for the first occurence of a specific feature in a unit of transcription and splitting the text at that point. The text before the split is treated as the speaker code, the part after the split is the transcribed text. The default is the colon-tab separator (`:  `). If the delimiter in your transcripts is inconsistent the extracted speaker codes will also be inconsistent.

Style information is not preserved: things like _italics_ and *bold* that are not present in the text itself will not be preserved. 

If you have a transcript where units of text have manual line breaks (as in the example below), each line will be treated as a separate unit. The below example will be treated as four separate units. In this case you can use the "assign previous speaker when no speaker identified option" to at least get the speaker codes right (at the cost of not being able to identify missing speaker codes). It will be more reliable to correct the source transcript first.

```
Speaker: This is text
with manual line breaks
Where what they said is
split.
```
