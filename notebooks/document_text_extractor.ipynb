# Document Text Extractor

This notebook takes a zip file containing a collection of PDFs and Microsoft Word files (`.docx` format only) and extracts the text from each document for further computational analysis.

It can be used to:

- Get the text of the your documents of interest ready for text analytics tools like [LDaCA Wordflow]() or corpus linguistic tools like [AntConc]().
- Make it easier to search and collate information across all your documents.
- Optionally link your documents back to the 

> NOTE: This notebook does not handle optical character recognition (OCR). If your documents are scans that don't include a searchable text overlay you will need to do additional work. If you need OCR we suggest starting with [Scribe OCR](https://scribeocr.com/) which also runs local on your own device in your browser.


## Usage

1. Collect your documents in `.docx` or `.pdf` format together and compress them into a single zip file ([instructions for Windows](https://support.microsoft.com/en-us/windows/zip-and-unzip-files-8d28fa72-f2f9-712f-67df-f80cf89fd4e5), [instructions for Mac](https://support.apple.com/en-au/guide/mac-help/mchlp2528/mac)). NOTE: other file types in your zip file will be ignored.
2. Run this notebook by hitting the ▶▶ ("restart kernel and run all cells") button in the menu above - you'll be asked if you want to restart this notebook - answer 'yes' and the notebook will be run.
3. Upload your zip file (`.zip`) file with the "Upload your documents" button.
4. (optional) Choose settings for how to prepare text, depending on the tools and analysis you want to conduct.
5. Generate your output dataset.
