# About the Output

## Formats

Extracting whole documents into text files will give you a `.zip` file containing one text file per original document uploaded. Each will be named after the source document with `.txt` appended so you can trace text files back to documents.

Extracting paragraphs/pages will produce a `.xslx` spreadsheet. Each row in the spreadsheet will correspond to one 'unit' of text: the page for PDFs and the paragraph or table for `.docx` files.

The included columns are:

- *source_file*: the uploaded file this unit was extracted from.
- *unit_type*: the unit type (page, paragraph or table).
- *sequence_number*: the occurrence of this unit in the text, starting from 0. For PDFs this is approximately the page number.
- *style*: for `.docx` documents the style associated with this paragraph. In a well structured document this may be useful in distinguishing different kinds of text.
- *text*: the text associated with this unit. All formatting is removed from the text.

## The Output Text Looks Weird Or Is Missing

Extracting text from rich documents (especially PDFs) is not straightforward. This tool makes a best effort using available libraries and trying to keep the configuration as simple as possible: it definitely won't handle every case appropriately. You will need to compare the extracted text with the original files to confirm that the outputs are reasonable. 

Failures can look like:

1. Text appears in the wrong order compared to the document.
2. Text is missing (for example if the PDF actually contains an image of text).
3. Text is strangely punctuated or the whitespace is inconsistent.
4. Letters appear jumbled.
5. "Too much" text is included (headers and footers may appear).
6. Text might be doubled up.
7. Whitespace is missing (for example: therearenospacesbetween words).
8. Formatting is missing (for example _italics_ or **bold** for emphasis).
